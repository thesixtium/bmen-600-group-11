## Analysing Data

```python

import numpy as np
import matplotlib.pyplot as plt
import mne
from scipy.signal import welch

X_all = np.asarray(X)
y_all = np.asarray(labels)
meta_all = meta.reset_index(drop=True)

fs = 250

channel_names = [
    "Fz",
    "FC3", "FC1", "FCz", "FC2", "FC4",
    "C5", "C3", "C1", "Cz", "C2", "C4", "C6",
    "CP3", "CP1", "CPz", "CP2", "CP4",
    "P1", "Pz", "P2", "POz"
]

classes = ["left_hand", "right_hand", "feet", "tongue"]

class_titles = {
    "left_hand": "Left Hand",
    "right_hand": "Right Hand",
    "feet": "Feet",
    "tongue": "Tongue"
}

bands = {
    "Mu (8–13 Hz)": (8, 13),
    "Beta (13–30 Hz)": (13, 30)
}

subjects = meta_all["subject"].to_numpy()
unique_subjects = np.sort(np.unique(subjects))

assert X_all.ndim == 3
assert X_all.shape[1] == len(channel_names)
assert len(X_all) == len(y_all) == len(subjects)

info = mne.create_info(
    ch_names=channel_names,
    sfreq=fs,
    ch_types="eeg"
)

info.set_montage(
    mne.channels.make_standard_montage("standard_1020")
)

freqs, psd_trials = welch(
    X_all,
    fs=fs,
    nperseg=min(500, X_all.shape[-1]),
    axis=-1
)

print("PSD shape:", psd_trials.shape)

subject_psd = {}

for subject in unique_subjects:

    subject_psd[subject] = {}

    for class_name in classes:

        mask = (
            (subjects == subject) &
            (y_all == class_name)
        )

        if not np.any(mask):
            raise ValueError(
                f"No trials for subject {subject}, "
                f"class {class_name}"
            )

        subject_psd[subject][class_name] = (
            psd_trials[mask].mean(axis=0)
        )

grand_psd = {}

for class_name in classes:

    grand_psd[class_name] = np.mean(
        np.stack([
            subject_psd[subject][class_name]
            for subject in unique_subjects
        ]),
        axis=0
    )

fig1, ax = plt.subplots(figsize=(13, 7))

mean_psd = np.mean(
    np.stack([
        np.mean(
            np.stack([
                subject_psd[subject][class_name]
                for class_name in classes
            ]),
            axis=0
        )
        for subject in unique_subjects
    ]),
    axis=0
)

for ch_idx, ch_name in enumerate(channel_names):

    mean_psd_db = 10 * np.log10(
        mean_psd[ch_idx] + 1e-20
    )

    ax.plot(
        freqs,
        mean_psd_db,
        linewidth=1.3,
        label=ch_name
    )

ax.axvspan(
    8, 13,
    color="green",
    alpha=0.12,
    label="Alpha / Mu (8–13 Hz)"
)

ax.axvspan(
    13, 30,
    color="orange",
    alpha=0.10,
    label="Beta (13–30 Hz)"
)

ax.set_xlim(0, 60)
ax.set_xlabel("Frequency (Hz)")
ax.set_ylabel("Power Spectral Density (dB/Hz)")

ax.set_title(
    "Grand-Average EEG Power Spectral Density\n"
    "Across 22 EEG Channels"
)

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    ncol=2,
    fontsize=9
)

ax.grid(alpha=0.25)

fig1.tight_layout()

plt.show()

subject_band_power_db = {}

for band_name, (fmin, fmax) in bands.items():

    freq_mask = (
        (freqs >= fmin) &
        (freqs <= fmax)
    )

    band_power_trials = np.trapezoid(
        psd_trials[:, :, freq_mask],
        freqs[freq_mask],
        axis=-1
    )

    subject_band_power_db[band_name] = {}

    for subject in unique_subjects:

        subject_band_power_db[band_name][subject] = {}

        for class_name in classes:

            mask = (
                (subjects == subject) &
                (y_all == class_name)
            )

            mean_power = band_power_trials[mask].mean(axis=0)

            power_db = 10 * np.log10(
                mean_power + 1e-20
            )

            subject_band_power_db[
                band_name
            ][subject][class_name] = power_db

grand_band_power_db = {}

for band_name in bands:

    grand_band_power_db[band_name] = {}

    for class_name in classes:

        grand_band_power_db[band_name][class_name] = np.mean(
            np.stack([
                subject_band_power_db[
                    band_name
                ][subject][class_name]
                for subject in unique_subjects
            ]),
            axis=0
        )

fig2, axes = plt.subplots(
    2, 4,
    figsize=(17, 8)
)

fig2.subplots_adjust(
    left=0.06,
    right=0.86,
    top=0.85,
    bottom=0.08,
    hspace=0.42,
    wspace=0.30
)

for row, band_name in enumerate(bands):

    values = np.concatenate([
        grand_band_power_db[band_name][class_name]
        for class_name in classes
    ])

    vmin = np.min(values)
    vmax = np.max(values)

    for col, class_name in enumerate(classes):

        ax = axes[row, col]

        power = grand_band_power_db[
            band_name
        ][class_name]

        im, _ = mne.viz.plot_topomap(
            power,
            info,
            axes=ax,
            cmap="viridis",
            vlim=(vmin, vmax),
            sensors=True,
            contours=5,
            show=False
        )

        ax.set_title(
            class_titles[class_name],
            fontsize=12
        )

    row_pos = axes[row, -1].get_position()

    cax = fig2.add_axes([
        0.90,
        row_pos.y0 + 0.02,
        0.012,
        row_pos.height - 0.04
    ])

    cbar = fig2.colorbar(im, cax=cax)
    cbar.set_label("Band Power (dB)", fontsize=10)

    axes[row, 0].text(
        -0.25,
        0.5,
        band_name,
        transform=axes[row, 0].transAxes,
        rotation=90,
        va="center",
        ha="center",
        fontsize=12
    )

fig2.suptitle(
    "Grand-Average Absolute EEG Band Power\n"
    "Across Four Motor Imagery Classes",
    fontsize=15
)

plt.show()

grand_differences = {}

for band_name in bands:

    grand_differences[band_name] = {}

    subject_differences = {
        class_name: [] for class_name in classes
    }

    for subject in unique_subjects:

        class_powers_linear = np.stack([
            10 ** (
                subject_band_power_db[
                    band_name
                ][subject][class_name] / 10
            )
            for class_name in classes
        ])

        reference_power = class_powers_linear.mean(axis=0)

        reference_db = 10 * np.log10(
            reference_power + 1e-20
        )

        for class_name in classes:

            class_db = subject_band_power_db[
                band_name
            ][subject][class_name]

            difference = class_db - reference_db

            subject_differences[class_name].append(difference)

    for class_name in classes:

        grand_differences[band_name][class_name] = np.mean(
            np.stack(subject_differences[class_name]),
            axis=0
        )

fig3, axes = plt.subplots(
    2, 4,
    figsize=(17, 8)
)

fig3.subplots_adjust(
    left=0.06,
    right=0.86,
    top=0.85,
    bottom=0.08,
    hspace=0.42,
    wspace=0.30
)

for row, band_name in enumerate(bands):

    values = np.concatenate([
        grand_differences[band_name][class_name]
        for class_name in classes
    ])

    limit = np.max(np.abs(values))

    if limit == 0:
        limit = 1e-12

    for col, class_name in enumerate(classes):

        ax = axes[row, col]

        difference = grand_differences[
            band_name
        ][class_name]

        im, _ = mne.viz.plot_topomap(
            difference,
            info,
            axes=ax,
            cmap="RdBu_r",
            vlim=(-limit, limit),
            sensors=True,
            contours=5,
            show=False
        )

        ax.set_title(
            class_titles[class_name] + " − All-Class Mean",
            fontsize=11
        )

    row_pos = axes[row, -1].get_position()

    cax = fig3.add_axes([
        0.90,
        row_pos.y0 + 0.02,
        0.012,
        row_pos.height - 0.04
    ])

    cbar = fig3.colorbar(im, cax=cax)
    cbar.set_label(
        "Log-Power Difference (dB)",
        fontsize=10
    )

    axes[row, 0].text(
        -0.25,
        0.5,
        band_name,
        transform=axes[row, 0].transAxes,
        rotation=90,
        va="center",
        ha="center",
        fontsize=12
    )

fig3.suptitle(
    "Grand-Average EEG Band Power Differences\n"
    "Reference: Mean Power Across All Four Classes",
    fontsize=15
)

plt.show()
```
